Minimal EDA for the GitHub issues dataset

This notebook loads the local Parquet file and gives a quick overview of the data shape, label balance, repository mix, and basic text characteristics.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('github-issues-dataset/github_issues_dataset.parquet')
df = pd.read_parquet(DATA_PATH)

print(df.head())
print('\nShape:', df.shape)
print('\nColumns:', list(df.columns))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Basic missing value check
print('Missing values per column:')
print(df.isna().sum())

# Priority and severity distributions
for col in ['priority', 'severity']:
    print(f'\n{col} distribution:')
    print(df[col].value_counts(dropna=False))

# Top repositories
print('\nTop 10 repositories by issue count:')
print(df['repo'].value_counts().head(10))

# Text length summaries
for col in ['title', 'body']:
    df[f'{col}_len'] = df[col].fillna('').astype(str).str.len()

print('\nTitle length summary:')
print(df['title_len'].describe())
print('\nBody length summary:')
print(df['body_len'].describe())

# Plot distributions
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.countplot(data=df, x='priority', ax=axes[0, 0], order=df['priority'].value_counts().index)
axes[0, 0].set_title('Priority Distribution')

sns.countplot(data=df, x='severity', ax=axes[0, 1], order=df['severity'].value_counts().index)
axes[0, 1].set_title('Severity Distribution')

sns.barplot(x=df['repo'].value_counts().head(10).index, y=df['repo'].value_counts().head(10).values, ax=axes[1, 0])
axes[1, 0].set_title('Top 10 Repositories')
axes[1, 0].tick_params(axis='x', rotation=45)

sns.histplot(df['body_len'], bins=40, ax=axes[1, 1], kde=True)
axes[1, 1].set_title('Body Length Distribution')
axes[1, 1].set_xlabel('Body Length (characters)')

plt.tight_layout()
plt.show()
```

## EDA findings (what actually matters for modeling)

- **114,073 issues, 68 repos** — but heavily skewed: pytorch, flutter, godot, rust, go alone are ~50% of the data. Any model will implicitly learn "pytorch-style" issues; that's a bias-audit item for Stage 6.
- **`priority` is extremely imbalanced**: low 101,937 / medium 10,132 / high 2,004 (≈ 89/9/2%). A model predicting "low" every time gets 89% accuracy — this is exactly why the brief demands **macro-F1**, which averages F1 over classes so the rare ones count equally.
- **`severity` is suspicious**: `Critical` is the *majority* class (66k of 114k). Real-world severity is never mostly-critical, so these labels are likely heuristically generated → this is our concrete example of **label noise** (Stage 2.3).
- **There is no `type` column.** The `labels` field holds raw repo labels (40,438 distinct strings — comma-joined lists like `bug,topic:editor` or `Suggestion,Awaiting More Feedback`). We must derive `type ∈ {bug, feature, question, docs}` ourselves — that's the taxonomy step below.
- **133 null bodies, 8 exact title+body duplicates** — minor, handled in cleaning.


# Step 2 — Label taxonomy: raw repo labels → `type`

**The problem:** every repo labels differently. `bug`, `Bug`, `c-bug`, `Issue-Bug`, `type: bug`, `bug-report` all mean the same thing; `enhancement`, `Suggestion`, `Proposal`, `feature request`, `c: new feature` all mean *feature*. There are 40k distinct label strings but only ~4,000 distinct *tokens* once we split the comma-joined lists — and a handful of keyword patterns cover most of them.

**The approach (and why):**
1. **Tokenize**: lowercase, split on commas → per-issue set of label tokens.
2. **Regex per class**: each of our 4 classes gets one pattern. We match per-token (not on the whole string) so word boundaries behave — e.g. `\bbug\b` matches `c-bug` and `bug-report` but *not* `debug` (a real vscode label!).
3. **Measure before deciding**: count how many issues match 0, 1, or 2+ classes. Issues matching 0 classes (e.g. only `triaged`, `P2`, `topic:editor`) get `type = None` — we *don't guess*; they're excluded from type training but still usable for the priority/severity tasks.
4. **Precedence for conflicts**: ~3% of issues match 2+ classes (e.g. `bug,documentation`). We resolve with a fixed precedence `docs > bug > feature > question` — docs first because it's the most specific signal, question last because it's the vaguest. This is a *documented modeling decision*, which is exactly what Stage 2.2/2.3 asks for.

This also settles **single-label vs multi-label**: the conflict rate is only ~3%, so single-label with documented precedence is a defensible simplification.


```python
import re

# One regex per target class. Order in PRECEDENCE resolves multi-class conflicts.
CLASS_PATTERNS = {
    'docs':     r'\bdocs?\b|documentation|typo',
    'bug':      r'\bbug\b|defect|\bcrash\b|regression|broken',
    'feature':  r'\bfeature\b|enhancement|suggestion|proposal|\brequest\b|improvement|\bidea\b',
    'question': r'question|\bfaq\b|q&a|discussion',
}
PRECEDENCE = ['docs', 'bug', 'feature', 'question']
compiled = {c: re.compile(p) for c, p in CLASS_PATTERNS.items()}

def label_classes(raw_labels: str) -> set:
    """Map a raw comma-joined label string to the set of matching classes."""
    tokens = [t.strip() for t in raw_labels.lower().split(',')]
    return {c for c, rx in compiled.items() for t in tokens if rx.search(t)}

class_sets = df['labels'].map(label_classes)

# Quantify before committing: how ambiguous is the mapping?
n_matches = class_sets.str.len()
print('Issues matching 0 / 1 / 2+ classes:')
print(n_matches.value_counts().sort_index())
print(f'\nCoverage (>=1 class): {100 * (n_matches > 0).mean():.1f}%')
print(f'Conflicts (>=2 classes): {100 * (n_matches > 1).mean():.1f}%')

def resolve(classes: set):
    for c in PRECEDENCE:
        if c in classes:
            return c
    return None  # unmapped -> excluded from type training

df['type'] = class_sets.map(resolve)

print('\nFinal type distribution:')
print(df['type'].value_counts(dropna=False))
```

```python
# Never trust a mapping you haven't audited.
# 1) What are we FAILING to map? If a frequent token clearly means bug/feature/etc.,
#    the patterns need extending.
unmapped_tokens = (
    df.loc[df['type'].isna(), 'labels']
      .str.lower().str.split(',').explode().str.strip()
      .value_counts()
)
print('Top 25 tokens among UNMAPPED issues:')
print(unmapped_tokens.head(25))

# 2) What raw labels landed in each class? Spot-check for wrong matches.
for cls in PRECEDENCE:
    print(f'\n--- top raw label strings mapped to "{cls}" ---')
    print(df.loc[df['type'] == cls, 'labels'].value_counts().head(5))
```
